## 2 · Plug in an AI model

Three ways to give the bot a key, best first:

1. Deepnote → **Environment** → Environment variables, then restart the machine.
2. The **🔑 API keys** panel in the app's sidebar → Save to server.
3. `save_key()` from a code block, like the cell below.

| Provider | Variable |
|---|---|
| OpenAI | `OPENAI_API_KEY` |
| Gemini | `GEMINI_API_KEY` |
| Groq | `GROQ_API_KEY` |
| NVIDIA NIM | `NVIDIA_API_KEY` |
| Ollama | none — runs locally |

Never leave a real key in a notebook you share or duplicate.

## 3 · Poke at the parts individually
Useful when you are debugging retrieval.

````
from lucelec_rag_bot import load_chunks, build_index, retrieve_chunks, answer, appliance_cost

index = build_index(load_chunks())
print(len(index["chunks"]), "chunks indexed")

for h in retrieve_chunks("how much does an air conditioner cost to run", index):
    print(h["score"], h["source"])
````

````
out = answer("What appliance uses the most electricity?", index)
print(out["reply"])
print([h["id"] for h in out["hits"]])
````

````
appliance_cost(watts=1200, hours_per_day=6)
````
